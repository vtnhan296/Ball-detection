```python
from pathlib import Path
from SoccerNet.Downloader import SoccerNetDownloader


def find_project_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    marker_sets = [
        ('data', 'notebooks', 'outputs'),
        ('data', 'outputs'),
        ('detect_player', 'tracking'),  # legacy layout fallback
    ]

    for candidate in [start, *start.parents]:
        for markers in marker_sets:
            if all((candidate / marker).exists() for marker in markers):
                return candidate
    return start


PROJECT_ROOT = find_project_root()
RAW_DATA_DIR = PROJECT_ROOT / 'data' / 'raw'
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)
print('Project root:', PROJECT_ROOT)
print('Raw data dir:', RAW_DATA_DIR)

downloader = SoccerNetDownloader(LocalDirectory=str(RAW_DATA_DIR))
downloader.downloadDataTask(task='tracking', split=['train'])
# downloader.downloadDataTask(task='tracking-2023', split=['train'])
```

```python
from pathlib import Path
from zipfile import ZipFile

zip_path = RAW_DATA_DIR / 'tracking' / 'train.zip'

if not zip_path.exists():
    raise FileNotFoundError(f'Could not find downloaded archive: {zip_path}')

with ZipFile(zip_path) as archive:
    names = archive.namelist()
    extract_dir = RAW_DATA_DIR if any(name.startswith('tracking/') for name in names) else RAW_DATA_DIR / 'tracking'
    archive.extractall(extract_dir)

zip_path.unlink()
print('Extracted tracking train data to:', RAW_DATA_DIR / 'tracking' / 'train')
```
